## 1. Load the corpus

A language model learns patterns from examples. In this tutorial the examples are not labeled questions and answers: they are characters from a text file, and later the training target will be the next character. First, we need to read the corpus as a string.

`Path` represents a filesystem path. `read_text` opens the file, decodes its bytes as UTF-8, and returns one Python string. The explicit encoding makes the file format clear and avoids relying on machine-specific defaults.

```python
from pathlib import Path

# Notebook kernels can start in the repository root, gpt-1/, or this folder.
data_candidates = (
    Path.cwd() / "gpt-1" / "data" / "input.txt",
    Path.cwd() / "data" / "input.txt",
    Path.cwd().parent / "data" / "input.txt",
)
dataset_path = next((path.resolve() for path in data_candidates if path.is_file()), None)

if dataset_path is None:
    raise FileNotFoundError("Could not find gpt-1/data/input.txt from this notebook's working directory.")

text = dataset_path.read_text(encoding="utf-8")

print(f"Dataset file: {dataset_path.resolve()}")
print(f"Length in characters: {len(text):,}")
print("\nFirst 1,000 characters:\n")
print(text[:1_000])
```

### Read the matching Java and C versions

The code above has three jobs: name the file, read all of its text, then report the size and a preview. The other implementations do those same jobs, with each language's normal project structure. This notebook groups them in one executable cell.

**Java:** start at [`Main.java`](../java/src/main/java/learning/karpathy/gpt1/Main.java). Its `main` method calls `TinyShakespeareDataset.findDefaultPath` to locate the shared file, then `TinyShakespeareDataset.load` reads it. That separate class owns the text and the operations for counting and previewing it. `Files.readString(path, UTF_8)` is Java's library-level equivalent of Python's `Path.read_text`.

**C:** start at [`main.c`](../c/src/main.c). It calls `dataset_read`, declared in [`dataset.h`](../c/include/dataset.h) and implemented in [`dataset.c`](../c/src/dataset.c). C's standard library exposes lower-level steps: `fopen` opens the file, `fseek` and `ftell` find its size, `malloc` reserves memory, and `fread` copies the bytes. The caller later releases that memory with `free`.

The C program currently expects to start from `gpt-1/c`; the VS Code lesson task sets that working folder for you. Java searches from the repository root, `gpt-1/`, or its Java project folder, so the Run button works from the workspace too.

One detail to notice: Python `len` counts Unicode code points, Java `String.length` counts UTF-16 code units, and the C implementation counts bytes. Tiny Shakespeare uses ASCII characters, so these counts agree for this dataset.
