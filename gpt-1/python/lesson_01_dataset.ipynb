## 1. Load the corpus

A language model learns patterns from examples. In this tutorial the examples are not labeled questions and answers: they are characters from a text file, and later the training target will be the next character. First, we need to read the corpus as a string.

`Path` represents a filesystem path. `read_text` opens the file, decodes its bytes as UTF-8, and returns one Python string. The explicit encoding makes the file format clear and avoids relying on machine-specific defaults.

```python
from pathlib import Path

dataset_path = Path("gpt-1/data/input.txt")
text = dataset_path.read_text(encoding="utf-8")

print(f"Dataset file: {dataset_path.resolve()}")
print(f"Length in characters: {len(text):,}")
print("\nFirst 1,000 characters:\n")
print(text[:1_000])
```

### Read the matching Java and C versions

The code above has three jobs: name the file, read all of its text, then report the size and a preview. The other implementations do those same jobs, with each language's normal project structure.

**Java:** start at [`Main.java`](../java/src/main/java/learning/karpathy/gpt1/Main.java). Its `main` method chooses `../data/input.txt` relative to the Java project folder and calls `TinyShakespeareDataset.load`. That separate class owns the text and the operations for counting and previewing it. `Files.readString(path, UTF_8)` is Java's library-level equivalent of Python's `Path.read_text`.

**C:** start at [`main.c`](../c/src/main.c). It calls `dataset_read`, declared in [`dataset.h`](../c/include/dataset.h) and implemented in [`dataset.c`](../c/src/dataset.c). C's standard library exposes lower-level steps: `fopen` opens the file, `fseek` and `ftell` find its size, `malloc` reserves memory, and `fread` copies the bytes. The caller later releases that memory with `free`.

The paths differ because each program is launched from its own project folder. Run the Java command from `gpt-1/java` and the C executable from `gpt-1/c`; both then reach the shared corpus with `../data/input.txt`.

One detail to notice: Python `len` counts Unicode code points, Java `String.length` counts UTF-16 code units, and the C implementation counts bytes. Tiny Shakespeare uses ASCII characters, so these counts agree for this dataset.
